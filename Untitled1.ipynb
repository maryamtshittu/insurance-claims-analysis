```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

claims = pd.read_csv('/claims-warehouse-v4-csv-claims.csv')
dates = pd.read_csv('/claims-warehouse-v4-csv-dates.csv')
policies = pd.read_csv('/claims-warehouse-v4-csv-policies.csv')
regions = pd.read_csv('/claims-warehouse-v4-csv-regions.csv')
product_lines = pd.read_csv('/claims-warehouse-v4-csv-product_lines.csv')

print("All datasets loaded successfully!")
print("Claims:", claims.shape)
print("Dates:", dates.shape)
print("Policies:", policies.shape)
print("Regions:", regions.shape)
print("Product lines:", product_lines.shape)
```

```python
# 1. Check whether the IDs link correctly
links = {
    "Policies": (claims["policy_id"], policies["id"]),
    "Product lines": (claims["product_line_id"], product_lines["id"]),
    "Regions": (claims["region_id"], regions["id"]),
    "Dates": (claims["date_id"], dates["id"])
}

for name, (used_ids, available_ids) in links.items():
    unmatched = set(used_ids) - set(available_ids)
    print(name, "unmatched IDs:", len(unmatched))


# 2. Check whether calendar information is correct
actual_dates = pd.to_datetime(dates["date"], errors="coerce")

print("\nIncorrect months:", (dates["month"] != actual_dates.dt.month).sum())
print("Incorrect years:", (dates["year"] != actual_dates.dt.year).sum())
print("Incorrect weekdays:", (dates["day_of_week"] != actual_dates.dt.day_name()).sum())


# 3. Check policy date consistency
start = pd.to_datetime(policies["start_date"], errors="coerce")
end = pd.to_datetime(policies["end_date"], errors="coerce")

print("\nPolicies ending before starting:", (end < start).sum())
```

```python
# Display 10 examples of incorrect date information
date_check = dates.copy()

actual = pd.to_datetime(date_check["date"])

date_check["actual_month"] = actual.dt.month
date_check["actual_year"] = actual.dt.year
date_check["actual_weekday"] = actual.dt.day_name()

date_check[
    (date_check["month"] != date_check["actual_month"]) |
    (date_check["year"] != date_check["actual_year"]) |
    (date_check["day_of_week"] != date_check["actual_weekday"])
].head(10)
```

```python
# Display 10 policies with inconsistent dates
policies[
    pd.to_datetime(policies["end_date"]) <
    pd.to_datetime(policies["start_date"])
].head(10)
```

```python
dates_clean = dates.copy()

actual_dates = pd.to_datetime(dates_clean["date"])

dates_clean["month"] = actual_dates.dt.month
dates_clean["year"] = actual_dates.dt.year
dates_clean["day_of_week"] = actual_dates.dt.day_name()

dates_clean.head()
```

```python
invalid_policies = policies[
    pd.to_datetime(policies["end_date"]) <
    pd.to_datetime(policies["start_date"])
]

invalid_policies.head(10)
```

```python
policies_clean = policies.copy()

policies_clean["valid_dates"] = (
    pd.to_datetime(policies_clean["end_date"]) >=
    pd.to_datetime(policies_clean["start_date"])
)

policies_clean["valid_dates"].value_counts()

analysis = claims.merge(
    policies_clean[["id", "coverage_type", "valid_dates"]],
    left_on="policy_id",
    right_on="id",
    suffixes=("", "_policy")
)

analysis = analysis.merge(
    regions[["id", "region_name"]],
    left_on="region_id",
    right_on="id",
    suffixes=("", "_region")
)

analysis = analysis.merge(
    product_lines[["id", "name"]],
    left_on="product_line_id",
    right_on="id"
)

analysis = analysis.merge(
    dates_clean[["id", "date"]],
    left_on="date_id",
    right_on="id",
    suffixes=("", "_date")
)

analysis = analysis.rename(columns={"name": "product_line"})

print("Combined dataset:", analysis.shape)
analysis.head()
```

```python
analysis["claim_date"] = pd.to_datetime(analysis["claim_date"])
analysis["date"] = pd.to_datetime(analysis["date"])

print(
    "Claims with mismatched dates:",
    (analysis["claim_date"] != analysis["date"]).sum()
)

analysis.loc[
    analysis["claim_date"] != analysis["date"],
    ["id_x", "claim_date", "date", "date_id"]
].head(10)
```

```python
# Overall claim amounts
print("CLAIM AMOUNTS")
print(analysis["claim_amount"].describe())

# Claim status
print("\nCLAIM STATUS")
print(analysis["status"].value_counts())

# Claims by product line
print("\nPRODUCT LINE ANALYSIS")
print(
    analysis.groupby("product_line")["claim_amount"]
    .agg(["count", "mean", "sum"])
    .round(2)
)

# Resolution times
print("\nRESOLUTION TIMES")
print(analysis["resolution_time_days"].describe())
```

```python
# Chart 1: Number of claims by product line

claims_by_product = analysis["product_line"].value_counts()

claims_by_product.plot(
    kind="bar",
    title="Number of Insurance Claims by Product Line",
    xlabel="Product Line",
    ylabel="Number of Claims",
    figsize=(9, 5)
)

plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()


# Chart 2: Average claim amount by product line

average_claims = analysis.groupby("product_line")["claim_amount"].mean()

average_claims.plot(
    kind="bar",
    title="Average Claim Amount by Product Line",
    xlabel="Product Line",
    ylabel="Average Claim Amount (£)",
    figsize=(9, 5)
)

plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

```python
regional_analysis = analysis.groupby("region_name").agg(
    number_of_claims=("claim_amount", "count"),
    average_claim=("claim_amount", "mean"),
    total_claim_value=("claim_amount", "sum")
).round(2)

regional_analysis = regional_analysis.sort_values(
    by="number_of_claims",
    ascending=False
)

regional_analysis
```

```python
regional_analysis["number_of_claims"].plot(
    kind="bar",
    title="Number of Insurance Claims by Region",
    xlabel="Region",
    ylabel="Number of Claims",
    figsize=(11, 5)
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    analysis["claim_amount"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Insurance Claim Amounts")
plt.xlabel("Claim Amount (£)")
plt.ylabel("Number of Claims")

plt.tight_layout()
plt.show()
```

```python
resolution_analysis = analysis.groupby("status").agg(
    number_of_claims=("id_x", "count"),
    average_resolution_days=("resolution_time_days", "mean"),
    median_resolution_days=("resolution_time_days", "median")
).round(2)

resolution_analysis
```
